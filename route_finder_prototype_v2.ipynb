```python
from osmnx_utils import initialize_graph
from route_plot_utils import RoutePlotter as RP
```

```python
origin = "Praça da Sé, São Paulo"
destinations = [
    ("Edifício Copan, São Paulo", 1),
    ("Mercado Municipal de São Paulo", 2),
    ((-23.5465, -46.6367), 3) 
]
```

```python
g_proj, route_nodes = initialize_graph(origin, destinations)
```

```python
import random
random_nodes = route_nodes[:1] + random.sample(route_nodes[1:], 3)
print(random_nodes)
```

```python
rp = RP(g_proj)
weight_function = rp.get_weight_function()
```

```python
segments_info = rp.compute_route_segments_info(random_nodes, weight_function, "priority")
```

```python
rp.plot_route_segments_info(segments_info)
```
