```python
import osmium
import geopandas
import pandas as pd
```

```python
fp = osmium.FileProcessor('./data/norte-260222.osm.pbf')\
         .with_locations()\
         .with_filter(osmium.filter.EntityFilter(osmium.osm.WAY))\
         .with_filter(osmium.filter.KeyFilter('highway'))\
         .with_filter(osmium.filter.GeoInterfaceFilter(tags=['highway', 'name', 'maxspeed', 'surface']))
```

```python
features = geopandas.GeoDataFrame.from_features(fp)
features.head()
```
