```python
import osmium
import geopandas
import pandas as pd
```

```python
fp = osmium.FileProcessor('./data/centro_sp_nano.osm')\
         .with_locations()\
         .with_filter(osmium.filter.EntityFilter(osmium.osm.WAY))\
         .with_filter(osmium.filter.KeyFilter('highway'))\
         .with_filter(osmium.filter.GeoInterfaceFilter())
```

```python
features = geopandas.GeoDataFrame.from_features(fp)
features.head()
```

```python
#exibir conteudo da primeira celula do primeiro registro
for i in range(10):
    print(features.iloc[i, 0])
```
